# 10.8 Python Errors

## Syntax Error

Compile (Transation) error, that is caught before the program runs. In other words, it stops the prorgam from being able to run.

## Exception / Runtime Error

The syntax is fine, so the program runs, however, because of this Exception, the program will stop and give you an error

## Logic Error

Type of runtime error. The program runs normally gives you output as it is suppose to. The problem is, if you aren't paying attention, the output is INCORRECT. Ex. print(2 + 2). OUTPUT: 28

Logic Errors can be difficult to spot. This is why you have to have an understanding of what the output should be.

# 10.9 Alorith Design

## Algorithm

A recipie for a solution

## Algorithms should be:

### Unambigious

Percise instructions at every step. No guesswork or personal opinion.

Ex. "Find the best-looking person."

### Executable

Every step can actually be carried out in practice.

Ex: "Use next decade's real interest rate."

### Terminating

The steps eventually come to an end.

Ex: "Repeat forever"

### How do we describe an algorithm in psudeo code?

1. Input & Outputs - What goes in? What do we want to come out?

2. Break it down - split the larger problem into smaller, executable sub tasks

3. Write psudeocode - Order the steps so values exist before they're used

4. Test it - Work a problem by hand with sample values.
